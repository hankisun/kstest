```python
import pandas as pd
df = pd.read_excel("D:\\myProjects\\exceltocsv\\exceltocsv.xlsx", sheet_name=0)
df
```

```python
print(range(len(df.index)))
```

```python
gbm=[]
gbm_list = ['A','B','C']
col = ['gbm','dept','index1','index2','index3']
prev = ""
for r in range(len(df.index)):
    if r==0 :
        prev = df.iloc[r,0]
    elif df.iloc[r,0] in gbm_list:
        prev = df.iloc[r,0]
    gbm.append(prev)
df.insert(0,'gbm',gbm)
df.columns = col
#print(df)
df1 = df.iloc[list(df['dept']!=df['gbm'])]
df2 = df1.drop(df1.tail(1).index)
df2.reset_index(drop=True, inplace=True)
print(df2)
```
