```python
from dataclasses import dataclass

@dataclass
class Item:
    price: int
    name: str

class DiscountStrategy:
    def discount(self):
        pass

class FirstGuestDiscount(DiscountStrategy):
    def __init__(self):
        self.rate = 0.8
    def discount(self, price):
        return price * self.rate
    
class NonFreshItemDiscount(DiscountStrategy):
    def __init__(self):
        self.rate = 0.7
    def discount(self, price):
        return price * self.rate
    
class Calculator:
    def __init__(self, strategy=None):
        self.discount_strategy = strategy
        self.items = []
    def add_item(self, item:Item):
        self.items.append(item)
    def calculate(self):
        sum_price = 0
        for item in self.items:
            sum_price += item.price
        if self.discount_strategy:
            sum_price = self.discount_strategy.discount(sum_price)
        return sum_price

cal = Calculator(FirstGuestDiscount())
cal.add_item(Item(price=500, name='snack'))
cal.add_item(Item(price=1000, name='fruit'))
cal.calculate()
```
