```python
from abc import ABC, abstractmethod

class ItemType():
    AGED_BRIE = "Aged Brie"
    BACKSTAGE_PASS = "Backstage passes to a TAFKAL80ETC concert"
    SULFURAS = "Sulfuras, Hand of Ragnaros"

class ItemFactory:
    @classmethod
    def create(cls, name, sell_in, quality):
        if name == ItemType.AGED_BRIE:
            return AgedBrie(name, sell_in, quality)
        elif name == ItemType.BACKSTAGE_PASS:
            return BackStagePass(name, sell_in, quality)
        elif name == ItemType.SULFURAS:
            return Sulfuras(name, sell_in, quality)
        else:
            return Normal(name, sell_in, quality)

class Item(ABC):
    def __init__(self, name, sell_in, quality):
        self.name = name
        self.sell_in = sell_in
        self.quality = quality

    def __repr__(self):
        return "%s, %s, %s" % (self.name, self.sell_in, self.quality)

    @abstractmethod
    def update_quality(self):
        pass

class AgedBrie(Item):
    def __init__(self, name, sell_in, quality):
        super().__init__(name, sell_in, quality)

    def update_quality(self, quality):
        print('quality of AgedBrieItem')
        self.quality = quality

class BackStagePass(Item):
    def __init__(self, name, sell_in, quality):
        super().__init__(name, sell_in, quality)

    def update_quality(self):
        print('quality of backstage pass')

class Sulfuras(Item):
    def __init__(self, name, sell_in, quality):
        super().__init__(name, sell_in, quality)

    def update_quality(self):
        print('quality of Sulfurass')

class Normal(Item):
    def __init__(self, name, sell_in, quality):
        super().__init__(name, sell_in, quality)

    def update_quality(self):
        print('quality of Normal')

```

```python
a = ItemFactory.create("Aged Brie", 10, 20)
```

```python
a.update_quality(30)
```
