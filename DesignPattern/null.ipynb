```python
class Bill:
    def __init__(self):
        self.price = []
        self.note = []
        self.total_price = 0.0
    def add_item(self, price, note='unknown'):
        self.price.append(price)
        self.note.append(note)
    def sum(self):
        return sum(self.price)
    def set_discount_rate(self, rate):
        self.rate = rate

class SpecialDiscount:
    def __init__(self):
        self.price_rate = 1.0
    def add_detail_to(self, bill):
        bill.set_discount_rate(self.price_rate)

class NewCustomerSpecialDiscount(SpecialDiscount):
    def __init__(self):
        self.price_rate = 0.9

class VIPCustomerSpecialDiscount(SpecialDiscount):
    def __init__(self):
        self.price_rate = 0.7
        
class LongCustomerSpecialDiscount(SpecialDiscount):
    def __init__(self):
        self.price_rate = 0.8        

class NullSpecialDiscount(SpecialDiscount):
    pass

class SpecialDiscountFactory:
    @classmethod
    def create(cls, customer):
        if cls.check_vip_customer(customer):
            return VIPCustomerSpecialDiscount()
        elif cls.check_long_customer(customer):
            return LongCustomerSpecialDiscount()
        elif cls.check_new_customer(customer):
            return NewCustomerSpecialDiscount()
        
        return NullSpecialDiscount()
    
    @classmethod
    def check_new_customer(cls, customer):
        if customer.is_new():
            return True
        return False
    
    @classmethod
    def check_vip_customer(cls, customer):
        if customer.is_vip():
            return True
        return False
    
    @classmethod
    def check_long_customer(cls, customer):
        if customer.is_long():
            return True
        return False
    
class Customer:
    def __init__(self, name):
        self.name = name
        self.vip = False
        self.long = False
        self.new = False

    def set_vip(self):
        self.vip = True
    def set_long(self):
        self.long = True
    def set_new(self):
        self.new = True
    def is_vip(self):
        return self.vip
    def is_long(self):
        return self.long
    def is_new(self):
        return self.new
    
def create_bill(customer):
    bill = Bill()
    bill.add_item(100, 'apple')
    bill.add_item(200, 'peach')
    bill.add_item(300, 'snack')

    special_discount = SpecialDiscountFactory.create(customer)
    special_discount.add_detail_to(bill)

    return bill

customer = Customer('ks')
customer.set_vip()
bill = create_bill(customer)
print(bill.rate)
print(bill.price)
```
